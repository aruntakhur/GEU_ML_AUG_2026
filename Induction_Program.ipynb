# Exercise

# Step 1: Load the Digits Dataset

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
import numpy as np

digits = load_digits()

X = digits.data
y = digits.target

print("Number of images:", len(X))
print("Image size:", digits.images[0].shape)
print("Classes:", np.unique(y))
```

# Step 2: Show Random Digits

```python
plt.figure(figsize=(12,4))

for i in range(10):
    idx = np.random.randint(0, len(X))
    plt.subplot(2,5,i+1)
    plt.imshow(digits.images[idx], cmap='gray')
    plt.title("Label = {}".format(y[idx]))
    plt.axis('off')

plt.show()
```

# Step 3: Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
```

# Step 4: Train Multiple Classifiers

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

models = {
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "KNN": KNeighborsClassifier(n_neighbors=3),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "SVM": SVC(),
    "Naive Bayes": GaussianNB()
}
```

# Step 5: Compare Accuracy

```python
from sklearn.metrics import accuracy_score

print("="*45)
print("{:<22}{}".format("Classifier","Accuracy"))
print("="*45)

for name, model in models.items():

    model.fit(X_train,y_train)

    pred = model.predict(X_test)

    acc = accuracy_score(y_test,pred)

    print("{:<22}{:.2f}%".format(name,acc*100))
```

# Step 6: Show AI Predictions

```python
idx = np.random.randint(0,len(X_test))

plt.imshow(X_test[idx].reshape(8,8), cmap='gray')
plt.axis('off')
plt.show()

print("Actual Label:", y_test[idx])
print()

for name, model in models.items():
    prediction = model.predict([X_test[idx]])
    print("{:<22}{}".format(name,prediction[0]))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

best_model = SVC()
best_model.fit(X_train,y_train)

ConfusionMatrixDisplay.from_estimator(best_model, X_test, y_test)
plt.show()
```
